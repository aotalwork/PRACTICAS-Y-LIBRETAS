### <font size=3> <i class="fa fa-pencil-square-o" aria-hidden="true"></i> Ejercicio 9</font>  

El archivo `data/title.ratings.tsv.gz` contiene datos sobre las votaciones del público en IMDb. Leer el archivo y cargarlo en un *DataFrame* denominado `df_ratings`. Fijar como índice la primera columna. 

Las columnas de este archivo están separadas por tabulaciones (caracter `"\t"`), y los valores perdidos se representan con `'\N'`. La función `pd.read_csv()` de _Pandas_ también leer directamente archivos comprimidos directamente. Para ello, en este caso, hay que pasarle el parámetro `compression='gzip'` (la lectura requiere varios segundos).

### <font size=3> <i class="fa fa-pencil-square-o" aria-hidden="true"></i> Ejercicio 16</font>  

En el *DataFrame* `df_people`, convertir la columna `gender` a categórica, utilizando las etiquetas especificadas en el diccionario `genres`.

```python
genres = {
    0: 'Not specified',
    1: 'Female',
    2: 'Male',
    3: 'Non-binary'
}

cat = df_people['gender'].map(genres)
df_people['gender'] = pd.Categorical(cat, categories=genres.values(), ordered=False)
df_people['gender']
```

### <font size=3> <i class="fa fa-pencil-square-o" aria-hidden="true"></i> Ejercicio 17</font> 
En el *DataFrame* `df_credits`, dejar solamente aquellas entradas correspondientes a las películas que contiene `df_movies`.

```python
df_credits=df_credits[df_credits["movie_id"].isin(df_movies.index)] ### COMPLETAR
df_credits
```

### <font size=3> <i class="fa fa-pencil-square-o" aria-hidden="true"></i> Ejercicio 18</font> 
Crear una *Series* denominada `directors`que cuyo índice sea `people_id` y que contenga el número de películas dirigidas por cada persona.

```python
df_directors = df_credits[df_credits["rol"] == "director"]
directors = df_directors.groupby("people_id")["movie_id"].count()
directors.name = "Numero de peliculas dirigidas" 
directors.index.name = "Director id"
directors.sort_values(ascending=False)
```

Crear una *Series* denominada `cast`que cuyo índice sea `people_id` y que contenga el número de películas en las que ha actuado cada persona.

```python
df_cast = df_credits[df_credits["rol"] == "cast"]
cast = df_cast.groupby("people_id")["movie_id"].count()
cast.name = "Numero de peliculas en las que han actuado" 
cast.index.name = "Cast id"
cast.sort_values(ascending=False)
```

### <font size=3> <i class="fa fa-pencil-square-o" aria-hidden="true"></i> Ejercicio 19</font> 

En el *DataFrame* `df_people`, crear una columna denominada `directs` que contenga la serie `directors`, y otra denominada `acts` que contengales la serie `cast`.

```python
### COMPLETAR

df_people["directs"] = directors
df_people["acts"] = cast

df_people
```

Mostrar los diez directores que más películas han dirigido

```python
### COMPLETAR
df_people.sort_values(by='directs',ascending=False).head(10)
```

Mostrar los diez actores/actrices que han participado en más películas.

```python
### COMPLETAR
df_people.sort_values(by='acts',ascending=False).head(10)
```

### <font size=3> <i class="fa fa-pencil-square-o" aria-hidden="true"></i> Ejercicio 20</font> 

¿Qué actores han participado en las peliculas que ha dirigido Clint Eastwood? Devolver el resutado ordenado por popularidad.

```python
### COMPLETAR
people = df_people[df_people['name']=='Clint Eastwood']
pelis_de_clint = df_credits[(df_credits["people_id"].isin(people.index))&(df_credits["rol"]=='director')]["movie_id"]
cast = df_credits[(df_credits["movie_id"].isin(pelis_de_clint))&(df_credits["rol"]=='cast')]["people_id"]
personas_en_pelis_de_clint = df_people[df_people.index.isin(cast)]
personas_en_pelis_de_clint.sort_values(by='popularity',ascending=False)
```
