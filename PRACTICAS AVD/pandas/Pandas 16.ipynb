## <font color="#5E5E5E">1. Marcas de tiempo</font>


### <font color="#5E5E5E"> El tipo de datos `TimeStamp`  </font>
Los objetos de tipo `Timestamp` representan marcas de tiempo que puede tener o no una zona horaria asociada. pandas admite distintas resoluciones temporales: segundos, milisegundos, microsegundos y nanosegundos. En pandas 3.0, al interpretar cadenas se utiliza normalmente la resolución de microsegundos, salvo que la entrada requiera nanosegundos. La resolución de almacenamiento no cambia la fecha y hora que estamos representando. [**( + inf )**](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.Timestamp.html).
 
 
La siguiente celda crea, mediante el constructor `pd.Timestamp()`, una marca de tiempo a partir de varios campos. Estos se pueden pasar también a través de argumentos posicionales, y no todos son necesarios.

```python
import pandas as pd

momento = pd.Timestamp(year=2026, month=10, day = 28, hour = 17, minute = 5, second = 30)
#momento = pd.Timestamp(2026, 10, 28, 17, 5, 30)    # Similar a la anterior, pero con argumentos posicionales.
#momento = pd.Timestamp(2026, 10, 28)               # Solo los obligatorios. No incluye hora. Toma el primer instante del día. 

print(momento," ",type(momento))
```

También se puede construir una marca de tiempo mediante un número, especificando la unidad de tiempo (desde el primer instante del 01/01/1970 en la zona UTC) que representa. 

```python
pd.Timestamp(1791895355.5, unit='s')            # Segundos, sin zona horaria asociada
#pd.Timestamp(1733835355.5, unit='s', tz='UTC') # Si se quiere asociar una zona horaria explícitamente
```

El constructor también permite crear `Timestamp` a partir de *strings*. Esta funcionalidad es muy útil ya que, como puede apreciarse en el fragmento de código, *pandas* acepta multitud de formatos, y los procesa de manera *"transparente"*.

```python
momento = pd.Timestamp('10/20/2026 6:05PM')
display(momento)
```

```python
momento = pd.Timestamp('Oct 20 2026 19:05')
display(momento)
```

```python
ahora = pd.Timestamp.now()
#ahora = pd.Timestamp.today() # Similar
display(ahora)
```

#### <font color="#5E5E5E"> La función `Timestamp.strftime()`  </font>


El método `Timestamp.strftime()` convierte una marca de tiempo en una cadena con el formato indicado. Por ejemplo, `%d/%m/%Y` muestra día, mes y año. Este cambio afecta a la representación textual: el objeto temporal original conserva su fecha y hora. Se puede encontrar información completa sobre el formato en la documentación [**( + inf )**](http://strftime.org/).

```python
ahora = pd.Timestamp.now()
ahora
```

```python
print(ahora.strftime("%d/%m/%Y"))
```

```python
print(ahora.strftime("%d of %B. %Y"))
```

El nombre se muestra en inglés pero es posible mostrarlo cambiando la configuración regional del entorno *python*.

```python
import locale

locale.setlocale(locale.LC_TIME, 'es_ES.UTF-8') 
```

```python
print(ahora.strftime("%d de %B. %Y"))
```

Esta función es similar a la implementada en el módulo `time` de Python.

```python
import datetime

ahora = datetime.datetime(year=2026, month=1, day=15, hour=11, minute=15, second=19)
print(type(ahora))
print(ahora.strftime("%d/%m/%Y"))
print(ahora.strftime("%d de %B. %Y"))
```

#### <font color="#5E5E5E"> La función `pd.to_datetime()` I </font>

La función `pd.to_datetime()` permite construir objetos de tipo `Timestamp` del mismo modo que el constructor de este tipo de objetos, pero es mucho más potente y flexible. La referencia completa de esta función puede encontrarse en la documentación [**( + inf )**](https://pandas.pydata.org/docs/reference/api/pandas.to_datetime.html).

El siguiente código muestra algunos ejemplos de construcción de objetos `Timestamp` mediante esta función.

```python
# A partir de una marca de tiempo en formato String
momento = pd.to_datetime('02/2/2026 10:05AM')               
display(momento)
```

```python
# Una marca de tiempo como origen, y un intervalo de 21 días (D)
inicio_curso=pd.to_datetime(21, unit='D', origin='2026-1-1')
display(inicio_curso)
```

El siguiente caso es ilustrativo de una circunstancia que puede dar lugar a errores inesperados.

```python
momento = pd.to_datetime('3/2/2026 10:05AM')
print(momento.strftime("%d of %B. %Y"))
```

```python
momento = pd.to_datetime('3/2/2026 10:05AM', dayfirst=True)
print(momento.strftime("%d of %B. %Y"))
```

<div class="alert alert-block alert-danger">
    
<i class="fa fa-exclamation-triangle" aria-hidden="true"></i>
Hay que tener en cuenta que en el sistema anglosajón, el mes se suele colocar antes que el día. La fecha 3/2/2026, que en España se interpreta como tres de febrero, se interpreta por defecto como dos de marzo. Esta circunstancia es fuente común de errores. Por eso es conveniente usar siempre el parámetro `dayfirst=True`, que da  preferencia al día primero.
</div>

Si la función `pd.to_datetime()` se aplica a un `Series` de *strings*, entonces devuelve un `Series` con objetos on un `dtype` temporal, aquí `datetime64[us]`; al acceder a un valor no ausente se obtiene un `Timestamp`.

```python
# Cada marca codificada de una manera.
serie = pd.Series(['2 June 2023', 'Aug 29, 2024', '2025-06-26', '7/12/26', '02/2/2027 10:05AM'])

serie_dt = pd.to_datetime(serie, format='mixed', dayfirst=True) 
serie_dt
#type(serie_dt[0])
```

La función `pd.to_datetime()` permite convertir fechas en formatos no predefinidos que pueden describirse mediante la cadena `format`. Esta funcionalidad puede utilizarse para llevar a cabo conversiones en serie.

```python
# Estas no funcionan porque los elementos temporales están en en_US.UTF-8'
# serie_str = pd.Series(['Fecha de finalización: Sep 15, 2023 a las: 10:30 pm',
#                        'Fecha de finalización: Oct 20, 2024 a las: 4:30 pm',
#                        'Fecha de finalización: Nov 25, 2025 a las: 4:30 am'])

serie_str = pd.Series(['Fecha de finalización: sept. 15, 2023 a las: 10:30 p. m.',
                       'Fecha de finalización: oct. 20, 2024 a las: 4:30 p. m.',
                       'Fecha de finalización: nov. 25, 2025 a las: 4:30 a. m.'])

serie = pd.to_datetime(serie_str, format='Fecha de finalización: %b %d, %Y a las: %I:%M %p')
serie
```


### <font color="#5E5E5E"> Indexación con marcas de tiempo: `DatetimeIndex`  </font>

La clase `DatetimeIndex` implementa un gran número de funcionalidades que permiten la indexación con marcas de tiempo. Es decir, como un índice formado por elementos de tipo `Timestamp`. 

```python
serie = pd.Series(['Módulo 1','Módulo 2', 'Módulo 3', 'Módulo 4', 'Módulo 5'], 
                  index=[pd.Timestamp('2026-01-22'), pd.Timestamp('2026-02-05'),
                   pd.Timestamp('2026-03-19'), pd.Timestamp('2026-04-26'),
                   pd.Timestamp('2026-05-7')])
serie
```

```python
serie.index                  # Imprime el índice
```

<div class="alert alert-block alert-info">

<i class="fa fa-info-circle" aria-hidden="true"></i> Una de las funcionalidades interesantes que ofrece consiste en que no es necesario proporcionar la marca exacta de tiempo para acceder a un elemento.

```python
serie.loc['2026-03-19']   # Fecha completa
#serie.loc['2026-04']      # Año y mes
#serie.loc['2026']         # Año
```

#### <font color="#5E5E5E"> Slicing </font>

Los índices de tipo `DatetimeIndex` también permiten hacer _slicing_.

```python
serie['2026-02':'2026-04']
#serie.loc['2026-02':]        # Se puede especificar una marca parcialmente
#serie['2025':]               # La marca concreta no tiene por qué aparecer en la serie
```

#### <font color="#5E5E5E"> La función `pd.to_datetime()` II </font>


Cuando esta función se aplica a una colección (en lugar de a un elemento o a un objeto `Series`, como se ha visto anteriormente), genera un objeto `DatetimeIndex`.



```python
fechas_str = ['2 June 2023', 'Aug 29, 2024', '2025-06-26', '7/12/23', '02/2/2024 10:05AM']
fechas = pd.to_datetime(fechas_str, format='mixed') 
fechas
```

Esta distinción no es nada intuitiva.

| Entrada a `pd.to_datetime()` en estos ejemplos | Resultado |
|---|---|
| Un valor escalar | `Timestamp` |
| Una `Series` | `Series` temporal, conservando su índice |
| Una lista o array | `DatetimeIndex` |
| Un `DataFrame` con componentes de fecha | `Series` temporal, una fecha por fila |

#### <font color="#5E5E5E"> La función `pd.date_range()` </font>

`pd.date_range()` construye un `DatetimeIndex` de marcas de tiempo. Podemos indicar inicio y fin, o una de esas fechas y el número de marcas mediante `periods`. En estos casos, si no indicamos `freq`, la frecuencia es diaria (`'D'`).

```python
ahora = pd.Timestamp.now()
luego = pd.Timestamp("2027-01-26")
```

```python
pd.date_range(ahora,luego)
```

<div class="alert alert-block alert-info">
    
<i class="fa fa-info-circle" aria-hidden="true"></i> Existen numerosas posibilidades en relación a la frecuencia [**( + inf )**](https://pandas.pydata.org/pandas-docs/stable/user_guide/timeseries.html#timeseries-offset-aliases). Por ejemplo, la siguiente porción de código genera marcas de tiempo diarias, pero que solamente corresponden a días de lunes a viernes.
</div>

```python
ahora = pd.Timestamp.now()
luego = pd.Timestamp("2027-01-26")
print(pd.date_range(ahora,luego,freq="B"))
```

## <font color="#5E5E5E">2. Periodos de tiempo </font>

### <font color="#5E5E5E"> El tipo de datos `Period`  </font>

El tipo de datos `Period` permite representar un periodo (intervalo) de tiempo, caracterizado por un momento de inicio y un momento de fin. La duración no es arbitraria, sino que corresponde a una frecuencia concreta (año, mes, etc.). El valor corresponde al periodo, no al principio ni al final del mismo.

```python
periodo = pd.Period('02/2026')            # Periodo. Corresponde a febrero de 2026.
#periodo = pd.Period('02/2026', freq='M') # Equivalente
print(periodo.days_in_month)              # Días del mes correspondiente al periodo 
print(periodo.start_time)                 # Marca de tiempo en la que comienza el periodo
print(periodo.end_time)                   # Marca de tiempo en la que termina el periodo
print(periodo.is_leap_year)               # Si el año en el que está comprendido el periodo es bisiesto
periodo
```

### <font color="#5E5E5E"> Indexación con periodos de tiempo: `PeriodIndex`  </font>

*pandas* también implementa índices construidos con periodos de tiempo (no marcas). Sin embargo, se puede acceder a los elementos mediante *strings*.

```python
ts = pd.Series(['primero','segundo','tercero','cuarto'], 
               [pd.Period('2026/2'),pd.Period('2026/2'),
                pd.Period('2026/3'),pd.Period('2027/10')])
ts
```

```python
ts.loc['2026/2']                  # Dos periodos corresponden esta marca de tiempo
#ts.loc['2026/3':]                # Del segundo elemento en adelante, según el orden de los elementos
```

El periodo especificado para acceder a los datos tiene que comprender los periodos utilizados en el índice, y no al revés.

```python
ts.loc['2026']        # Devuelve los elementos comprendidos en 2020
#ts.loc['2026/2/15']  # Produce fallo
```

Si se pasa un objeto de tipo `pd.Period()` (no es una situación común) el objeto debe corresponder con la resolución del índice. No es demasiado intuitivo.

```python
#ts.loc[pd.Period('2026')]           # No funciona porque el índice está hecho con meses.
ts.loc[pd.Period('2026/2')]     
#ts.loc[pd.Period('2026/2/15')]      # No funciona porque el índice está hecho con meses.
```

<div class="alert alert-block alert-danger">
    
<i class="fa fa-exclamation-triangle" aria-hidden="true"></i>
Los elementos de un `PeriodIndex` comparten una frecuencia. Si se mezclan periodos mensuales y anuales en esta lista, el índice resultante será un `Index` genérico de dtype `object`. Se perderá el acceso temporal mediante cadenas que se acaba de utilizar.
</div>

```python
ts = pd.Series(['primero','segundo','tercero','cuarto'], 
               [pd.Period('2026/2'),pd.Period('2026/3'),
                pd.Period('2026/4'),pd.Period('2026')])

#ts.loc['2026/2']                 # No funciona
#ts.loc[pd.Period('2026-02', freq='M')]  # Funciona
#tslindex
```

## <font color="#5E5E5E">3. Intervalos de tiempo</font>


### <font color="#5E5E5E"> El tipo de datos `Timedelta`  </font>


Los objetos de tipo `Timedelta`, permiten representar intervalos de tiempo, caracterizados únicamente por una duración pero sin instantes de inicio y final, y operar con ellos.

```python
intervalo = pd.Timedelta('7days 3hours')
otro_intervalo = pd.Timedelta('2days 1hour 25minutes')

intervalo
```

```python
otro_intervalo
```

```python
intervalo+otro_intervalo
```

Algunas operaciones entre objetos `Timestamp` generan intervalos de tiempo (`Timedelta`).

```python
plazo = pd.Timestamp('Mar 10, 2026') - pd.Timestamp('3 February 2026')
plazo
#type(plazo)
```

También es posible operar entre marcas e intervalos de tiempo.

```python
deadline = pd.Timestamp('Jan 10, 2026 8:10AM') + pd.Timedelta('7D 3h')
deadline
#type(deadline)
```

### <font color="#5E5E5E"> Indexación con intervalos de tiempo: `TimedeltaIndex`  </font>

También se pueden construír índices a partir de objetos `Timedelta`. Estos se representan con objetos de tipo `TimedeltaIndex`.

```python
index = [pd.Timedelta('0:30:00'), pd.Timedelta('12:00:00'),
         pd.Timedelta('1D'), pd.Timedelta('1D 12:00:00')]

ts = pd.Series(['Media hora', 'Medio día', 'Día', 'Día y medio'], index )
ts
#ts.index
```

Cuando se usa un objeto `Timedelta` como clave de `.loc`, se debe usar una duración exacta del índice. Si no existe, se produce un error.

```python
ts.loc[pd.Timedelta('0:30:00')]
ts.loc[pd.Timedelta('1D 12:00:00')]
#ts.loc[pd.Timedelta('0:30:35')]   # No funciona
```

Se puede comparar intervalos de tiempo, lo que permite el acceso condicional.

```python
ts.loc[ts.index>pd.Timedelta('1 h')]
```

#### <font color="#5E5E5E"> La función `pd.timedelta_range()` </font>

La función `pd.timedelta_range()`, similar a las vistas anteriormente,  permite generar índices con intervalos de tiempo.

```python
pd.timedelta_range(start='1 day', periods=4)     # Duraciones de 1, 2, 3 y 4 días
#pd.timedelta_range(start='1 day', periods=4, closed="right")
#pd.timedelta_range(start='1 day', end='5 days', periods=4)
```

```python
pd.timedelta_range(start='1 day', end='2 days', freq='6h')  # El primer periodo es un día. Luego cada uno añade seis horas
                                                            # hasta que llega a un periodo de dos días. 
```

#### <font color="#5E5E5E"> La función `pd.to_timedelta()` </font>

Permite generar intervalos de tiempo a partir de cadenas de texto.El resultado depende del formato de la entrada

| Entrada  | Resultado |
|---|---|
| Una cadena | `Timedelta` |
| Una `Series` de cadenas | `Series` con dtype temporal |
| Una lista de cadenas | `TimedeltaIndex` |

```python
pd.to_timedelta('1D 12:00:0') # String (escalar)
```

```python
pd.to_timedelta(pd.Series(['1D 12:00:0', '20s', '345us']))
```

```python
pd.to_timedelta(['1D 12:00:0', '20s', '345us'])
```

```python
pd.to_timedelta(range(5), unit='D')
```
