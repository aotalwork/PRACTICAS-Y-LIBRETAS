## <font color="#5E5E5E">1. Acceso a columnas de un *DataFrame*</font>


El operador `[ ]` (en realidad es el método `__getitem__()`)  proporciona acceso a las columnas del `DataFrame`. Toma ***un** argumento* que puede ser:
* El nombre de una columna. En ese caso devuelve un objeto de tipo `Series`.
* *Una lista* con nombres de columnas únicos. En ese caso devuelve otro `DataFrame`.

En esta celda se accede, mediante el `string` con el nombre, al contenido de la columna `Producto` del `DataFrame` `df_productos`. El resultado es un objeto de tipo `Series`.

```python
s_productos = df_productos['Producto']
s_productos
```

El uso de una lista de *strings* proporciona acceso a un conjunto de columnas, y devuelve otro `DataFrame`.

```python
#columnas = ['Producto','Compra','Venta']
#df_productos[columnas]

df_productos[['Producto','Compra','Venta']]   # Equivalente 
```

La lista de nombres puede tener tamaño uno. En ese caso, se devuelve un `DataFrame` con una columna (en lugar de una *Serie*).

```python
df_productos[['Producto']]
```

```python
#df_productos['Producto','Compra','Venta']  # Devuelve error.
```

Por ejemplo: `df_productos.Producto`. Sin embargo, **es recomendable no utilizar este tipo de notación** o, sobre todo, no mezclar los tipos de anotación.

## <font color="#5E5E5E">2. Acceso a filas (elementos) mediante índice: `.loc[]`</font>


El acceso a filas de un `DataFrame`, o a los elementos de una `Series`, **se debe hacer** (están concebidos para ello) a través de su índice, mediante el método `DataFrame.loc[idx]`/`Series.loc[idx]`. Recibe un **único** argumento,  `idx`, que puede ser:

* un valor, identificador de un elemento del índice;
* una lista de valores, identificadores de elementos del índice.

<div class="alert alert-block alert-info">
    
<i class="fa fa-info-circle" aria-hidden="true"></i> En este caso se verá acceso mediante valores del índice. Más adelante se verán más posibilidades.
</div>


### <font color="#5E5E5E">  Acceso mediante un valor </font>


`DataFrame.loc[idx]`, cuando `idx` es un valor, devuelve la *fila o filas* correspondientes del `DataFrame`. Si el valor del índice corresponde a una sola fila, ésta se devuelve en forma de `Series`.

```python
df_productos
```

```python
df_productos.loc['Tienda 1']
```

Si el índice corresponde a varias filas **(puede ser)**, el resultado es, a su vez, otro `DataFrame`.

```python
df_productos.loc['Tienda 2']
```

En cuanto al acceso a `Series`, el resultado depende también del número de elementos correspondientes al índice. Cuando el índice corresponde a un elemento, se devuelve un escalar.

```python
s_productos.loc['Tienda 1']
```

Cuando el índice corresponde a varios elementos,  se devuelve otra `Series`.

```python
s_productos.loc['Tienda 2']
```



### <font color="#5E5E5E">  Acceso mediante una colección de valores </font>


`DataFrame.loc[idx]` acepta como argumento una *secuencia* de valores, y devuelve siempre un nuevo `DataFrame` con las filas seleccionadas.

```python
df_productos.loc[['Tienda 1', 'Tienda 2']]
#df_productos.loc[['Tienda 1']]
```

En el caso de `Series.loc[idx]`, cuando `idx` es una colección, el método devuelve siempre otra `Series`.

```python
s_productos.loc[['Tienda 1', 'Tienda 2']]
#s_productos.loc[['Tienda 1']]
```

### <font color="#5E5E5E">  <i>slicing</i></font>


El método `Dataframe.loc[idx]`/`Series.loc[idx]` también permite especificar el conjunto de valores `idx` mediante *slicing*, independientemente del tipo de datos utilizado en el índice. Por ejemplo, en la siguente celda obtiene las filas que van desde `Tienda 1` hasta `Tienda 3`.

```python
df_productos
```

```python
df_productos.loc['Tienda 1':'Tienda 3']
```

El funcionamiento es similar en *Series*.

```python
s_productos.loc['Tienda 1':'Tienda 3']
```

<div class="alert alert-block alert-danger">
    
<i class="fa fa-exclamation-triangle" aria-hidden="true"></i>
__Importante:__ Los elementos utilizados para definir los límites del intervalo deben ser *únicos* en el índice. Por otra parte, el  *slicing* incluye el límite derecho. 
</div>

En el ejemplo anterior se aprecia que, al hacer *slicing*, queda fuera una de las filas correspondientes a `Tienda 2` (la última fila). **Es muy importante ser consciente del orden que guardan las filas** (o de si no tienen por qué guardarlo) al hacer este tipo de acceso. Debido a esto, prácticamente siempre es necesario ordenar las filas del `DataFrame` en función del valor de su índice mediante el método `DataFrame.sort_index()`.

```python
df_productos.sort_index().loc['Tienda 1':'Tienda 3']
```

<div class="alert alert-block alert-info">
    
<i class="fa fa-info-circle" aria-hidden="true"></i> Los índices de cada tipo ofrecen funcionalidades específicas. En el caso de índices alfanuméricos, se puede hacer _slicing_ lexicográfico con los prefijos, sin necesidad de especificar los valores concretos de los elementos. Los índices de datos temporales, por ejemplo, ofrecen funcionalidades muy completas y complejas que se verán en el módulo correspondiente.
</div>

```python
df_productos.set_index('Producto').sort_index().loc['Az':'Vi']
#df_productos.set_index('Producto').sort_index().loc['Az':'Vino']
```

Una alternativa (poco usada) al *slicing* es `DataFrame.truncate()`, que requiere el eje ordenado y conserva los límites.

```python
df_productos.sort_index().truncate(before='Tienda 2', after='Tienda 3')
```

###  <font color="#5E5E5E"> Selección con etiquetas que pueden no existir.</font>

Cuando se seleccionan filas a partir de un conjunto de etiquetas, y alguna de las etiquetas del conjunto no corresponde a ninguna entrada del índice (y por tanto a ninguna fila), se produce una excepción. Si en lugar de eso simplemente se quieren descartar esas etiquetas, se puede hacer de dos formas. 

Si el índice del `DataFrame` original es único, resultado del método `DataFrame.reindex(labels)` en cierto modo, es equivalente a una selección. Otra opción consiste en utilizar una intersección : `DataFrame.loc[DataFrame.index.intersection(labels)]`.
</div>

```python
etiquetas = ['Tienda 1', 'Tienda 2', 'Tienda 6']
# df_productos.reindex(etiquetas) # Salta excepción porque `Tienda 2` aparece dos veces.
df_productos.loc[df_productos.index.intersection(etiquetas)]
```


###  <font color="#5E5E5E"> Iteración sobre filas</font>


El método `DataFrame.iterrows()` permite iterar sobre las filas de un `DataFrame`. Devuelve pares (índice, `Series`), de forma que la correpsondiente a cada fila tiene un único `dtype`.

```python
for idx, row in df_productos.iterrows():           
    print(idx, end=' --> ')                        
    print(row['Producto'])
```

El método `DataFrame.itertuples()` devuelve tuplas con nombres de campo. Es más eficiente que `DataFrame.iterrows()` y preserva los tipos de datos.

```python
for row in df_productos.itertuples(): 
    print(row)
    #print(row.Nombre, row.Producto)
```

##  <font color="#5E5E5E"> 3. Acceso a elementos en `DataFrames`</font>


Con `DataFrame.loc[idx, col]` se accede a los elementos dados el valor (o valores) de su índice y columnas. Tanto `idx` como `col` pueden ser un valor, o una colección de valores, que puede definirse mediante *slicing*.

```python
df_productos.loc[['Tienda 2','Tienda 1'],'Producto']
#df_productos.loc[['Tienda 2','Tienda 1'],['Producto']]
```

```python
df_productos.loc[['Tienda 2','Tienda 1'],['Producto','Compra','Venta']]
```

```python
df_productos.loc['Tienda 1':'Tienda 3',['Producto','Compra','Venta']]
```

<div class="alert alert-block alert-warning">
    
<i class="fa fa-exclamation-circle" aria-hidden="true"></i>
Es frecuente ver código en el que se accede a los elementos por índice, y se accede después a la columna resultante, es decir `DataFrame.loc[idx][col]` . Aunque en consultas el comportamiento es el esperado, *no se debe utilizar este tipo de acceso*, sobre todo si se va a escribir sobre `DataFrame` seleccionado. Esta circunstancia se tratará con detalle en una sección posterior.
</div>



### <font color="#5E5E5E">  El método `.at[]`</font>

El método  `Dataframe.at[idx, col]` es equivalente a `Dataframe.loc[idx,col]`, pero con una salvedad: solo admite escalares como argumentos, es decir, **no admite colecciones**. Sin embargo, es más eficiente. 

```python
df_productos.at['Tienda 1', 'Apellido']
```

##  <font color="#5E5E5E"> 4. Acceso para escritura</font>


El acceso  para escritura en `Series` y `DataFrame` es similar al acceso para lectura. Pero debe haber una correspondencia entre las posiciones accedidas y los valores asignados.

<div class="alert alert-block alert-warning">
    
<i class="fa fa-exclamation-circle" aria-hidden="true"></i>
Para ilustrar el siguiente ejemplo, se copiará el contenido de la columna `df_productos['Producto']` en la serie `s_productos`.  Debido a que *pandas* desde la versión 3.0 usa *copy-on-write*, la copia se hace de forma implícita cuando se produce una escritura en la `Serie`, de modo que no se altera el `DataFrame` original.
</div>


```python
s_productos = df_productos['Producto']
s_productos
```

```python
#df_productos
```

Si se asigna un escalar a varias posiciones de un objeto *Series*, se escribe en todas ellas.

```python
s_productos.loc[['Tienda 1','Tienda 3']]='Jamón'
s_productos
```

Cuando se asigna una colección de valores a varias posiciones de una `Series`, las dimensiones de ambos conjuntos han de ser similares (salvo en el caso en que se asignen escalares, como se ha visto).

```python
s_productos.loc['Tienda 1'] = 'Vermut'  
#s_productos.loc[['Tienda 1','Tienda 3']]=['Vermut', 'Chorizo']
#s_productos.loc[['Tienda 1','Tienda 3']]=['Vermut', 'Vermut', 'Vermut']      # Error por la diferencia de tamaños
#s_productos.loc[['Tienda 1','Tienda 2']]=['Vermut', 'Chorizo']                # Error por la diferencia de tamaños
s_productos
```

Si se asigna un valor a un nuevo índice, se añade un nuevo elemento a la serie.

```python
#s_productos.loc['Tienda 5']
```

```python
s_productos.loc['Tienda 5'] = 'Tomates'
s_productos
```

En el caso de los `DataFrame`, también se puede asignar un valor escalar a varios elementos, siempre y cuando el tipo de datos sea compatible.

```python
df_productos.loc[['Tienda 1','Tienda 2'],['Compra', 'Venta']]=10
# df_productos.loc[['Tienda 1','Tienda 2'],['Compra', 'Venta']]=10.5.  # Se produce excepción porque `Compra` es un entero.`
df_productos
```

Igualmente, cuando se asigna una colección de valores, el tamaño de conjunto de datos y de las posiciones deben coincidir.

```python
df_productos.loc[['Tienda 1','Tienda 2'],'Venta']=[22, 14.5, 0]
df_productos
```

También se puede escribir el valor de columnas completas. Al igual que anteriormente, han de coincidir los tamaños.

```python
df_productos['Producto'] = ['Queso manchego', 'Vino manchego', 
                            'Jamón extremeño', 'Aceite andaluz', 'Azafrán manchego']
df_productos
```

```python
df_productos['C.P'] = ['02001', '02001', '02003', '02004','02004']
df_productos
```

```python
df_productos['Localidad'] = 'Albacete'
df_productos
```

## <font color="#5E5E5E"> <i class="fa fa-plus-circle" aria-hidden="true"></i> 5. Acceso por posición `.iloc[]`, `.iat[]`.  </font>

Aunque el acceso a filas de un `DataFrame` o elementos de una `Series` **se debe hacer mediante el índice**, también es posible hacerlo por su posición, del mismo modo que se hace en el resto de colecciones de *python* o *numpy*.
<div class="alert alert-block alert-danger">
    
<i class="fa fa-exclamation-triangle" aria-hidden="true"></i> Ha de evitarse el acceso a los datos por posición a menos que sea necesario. No es una práctica prohibída, pero implica tener que controlar el orden, y es más fácil que se produzcan errores.
</div>

### <font color="#5E5E5E">  Acceso a filas y elementos por posición: `.iloc[]` </font>



Se puede acceder a las filas de un `DataFrame` o a los elementos de una `Series` mediante `DataFrame.iloc[pos]/Series.iloc[pos]`, donde `pos` es un número entero que representa la posición. El funcionamiento de `DataFrame.iloc[pos]` es similar al de `DataFrame.loc[idx]`. Toma **un argumento** que puede ser un valor entero (en este caso una posición), o una colección de enteros que puede representarse con *slicing*.  En el primer caso, devuelve una `Series` con la fila correspondiente.


```python
df_productos.iloc[0]
```

En caso de utilizar una colección, devuelve un `DataFrame` con las filas correspondientes. 

```python
df_productos.iloc[[1,3,4]]
```

El uso de *slicing* es equivalente al uso de una colección de enteros.

```python
df_productos.iloc[0:3]
```

En el caso de `Series.iloc[pos]`, cuando `pos` es un entero, devuelve elemento correspondiente.

```python
s_productos.iloc[4]
```

Cuando es una colección de enteros, devuelve la otra `Series` con los elementos. 

```python
df_productos.iloc[0:3]
```

El método `DataFrame.take()` es parecido a `DataFrame.iloc[]`.

```python
df_productos.take([1, 3, 4])
```



### <font color="#5E5E5E">  Acceso al elementos por posición</font>

También es posible acceder a elementos, o conjuntos de elementos, por posición mediante `DataFrame.iloc[pos,col]`, donde `pos` y `col` pueden ser tanto enteros como colecciones de enteros.

```python
df_productos.iloc[2,1]
```

```python
df_productos.iloc[0:3,0:2]
```

<br>

### <font color="#5E5E5E">  El método `.iat[]`</font>

El método `Dataframe.iat[idx, col]` permite acceder a elementos por posición, pero tanto `idx` como `col` han de ser valores enteros, y el valor devuelto es un escalar.



```python
df_productos.iat[2,1]    
```

## <font color="#5E5E5E"> <i class="fa fa-plus-circle" aria-hidden="true"></i> 6. Ejes y conversión a arrays </font>


Un `DataFrame` organiza los datos en filas y columnas. Se puede acceder a los dos índices (y sus etiquetas) mediante los atributos:

* `index`: contiene las etiquetas de las filas.
* `columns`: contiene las etiquetas de las columnas.

Las etiquetas de los índices ***deben distinguirse de los datos***. 

```python
df_productos.index
```

```python
df_productos.columns
```



### <font color="#5E5E5E">  Etiquetas como arrays y listas</font>


El atributo `array` de un `Index` permite acceder al array que contiene sus etiquetas. Devuelve un `ExtensionArray`: una estructura de pandas cuya implementación depende del tipo de los datos. Esta se puede convertir en una lista mediante el método `tolist()`.

```python
df_productos.columns.array
#df_productos.columns.tolist()
```

### <font color="#5E5E5E"> Conversión de `DataFrame` a arrays *NumPy*</font>


El método `to_numpy()` devuelve los datos como un array bidimensional de NumPy (`ndarray`). Conserva el orden de las filas y columnas, pero no incluye los índices (ni, por tanto, sus etiquetas). 

Un array de NumPy tiene un único `dtype`, por lo que pandas debe buscar un tipo común para todas las columnas. La conversión de tipos puede requerir copiar los datos y cambiar su representación. No modifica los tipos de las columnas del `DataFrame` original.

```python
datos_numpy = df_productos.to_numpy()

# Muestra las dos primeras filas.
datos_numpy[:2]
```

```python
datos_numpy.dtype
```

```python
# Al seleccionar solo las columnas numéricas, se obtiene un array de números decimales.
df_productos[['Compra', 'Venta']].to_numpy()
```

### <font color="#5E5E5E"> Conversión de `Series`</font>

La secuencia de valores de un objeto `Series` se puede obtener de dos maneras:

* `serie.array`: devuelve un `ExtensionArray` y conserva el tipo de datos de pandas.
* `serie.to_numpy()`: devuelve un `ndarray` de NumPy, lo que puede requerir una conversión y una copia.

En ambos casos se obtiene una estructura unidimensional, sin las etiquetas del índice. 

La distinción es especialmente útil con los tipos de texto de pandas 3.0. Su representación en pandas utiliza arrays específicos; al convertirlos a NumPy podemos obtener un array de tipo `object`. Por otra parte, la clase concreta que muestra `.array` puede variar según el tipo y el almacenamiento utilizado, por ejemplo, con o sin PyArrow.

También existe el atributo `.values`, tanto en `Series` como en `DataFrame`. Es público y se puede utilizar, pero preferimos `.array` o `.to_numpy()` porque expresan con mayor claridad qué representación necesitamos. Un `DataFrame` no dispone de un atributo `.array`.

<div class="alert alert-block alert-info">
    
<i class="fa fa-info-circle" aria-hidden="true"></i>  Tanto en `Series` como `DataFrame` disponen de un atributo público, `.values`, pero se recomiendan los métodos indicados. Un `DataFrame` no dispone de un atributo `.array`.
</div>

```python
s_texto = df_productos['Producto']

s_texto.array
```

```python
s_texto.to_numpy()
```

### <font color="#5E5E5E"> Arrays, copias y `Copy-on-Write`</font>



Un array de NumPy que comparte memoria con un objeto de pandas se devuelve como **solo lectura**. Así se evita modificar indirectamente el objeto original escribiendo en ese array.

Esto no ocurre con el atributo `.array` que respalda la serie y no crea una copia independiente. Modificarlo directamente puede eludir Copy-on-Write y afectar a los datos originales.

```python
df_numerico = pd.DataFrame({'Valor': [10, 20, 30]})

matriz = df_numerico.to_numpy()

# En este ejemplo devuelve False: el array es de solo lectura.
matriz.flags.writeable

# Descomentar esta asignación produciría un ValueError.
# matriz[0, 0] = 99
```

Si se necesita modificar los valores del array por separado, se puede producir una copia mediante `to_numpy(copy=True)`.

```python
matriz_copia = df_numerico.to_numpy(copy=True)
matriz_copia[0, 0] = 99

matriz_copia
```
