### <font color="#5E5E5E"> Lectura y escritura de archivos de texto</font>
La función `pd.read_csv()` lee un archivo de datos con formato `.csv ` (*comma separated values*) o similar, y lo almacena en un `DataFrame`. Recibe como argumento principal el nombre del archivo, que puede ser también una URL, y acepta una gran cantidad de parámetros [**( + inf )**](http://pandas.pydata.org/pandas-docs/stable/generated/pandas.read_csv.html). Algunos de los más importantes son:


* `index_col`, que permite especificar si alguna de las columnas ha de ser utilizada como índice del `DataFrame`;
* `names`, que debe contener los nombres de las columnas;
* `skiprows`, que indica qué líneas del archivo han de ser descartadas;
* `sep`, separador entre columnas. Por defecto es `','`, pero a menudo es `';'` o un tabulador;
* `dtype`, que permite especificar el tipo de datos de cada columna;
* `na_values`, que indica cómo se representan los valores perdidos en el archivo;
* `quotechar`, que indica qué carácter se utiliza para delimitar cadenas de texto; 
* `encoding`, con la codificación del archivo;
* etc.

<div class="alert alert-block alert-info">
    
<i class="fa fa-info-circle" aria-hidden="true"></i>
La guía de usuario de `pd.read_csv()` es muy completa [**(+ inf)**](https://pandas.pydata.org/pandas-docs/stable/user_guide/io.html#csv-text-files) .
</div>

En este ejemplo se construye un `DataFrame` a partir de `data/eukaryotes.tsv`. Es un archivo `.tsv`: cada línea representa una fila y sus campos están separados por tabuladores. La columna `Species` se utiliza como índice.

```python
import pandas as pd

df_sequences = pd.read_csv('data/eukaryotes.tsv', sep='\t', index_col='Species')
df_sequences.head(5)
```

En la siguiente celda se lee el archivo de datos `titanic.csv` desde una URL, descartando la cabecera, y asignando a las columnas los nombres en español en la lista `names`. Además, se establece como índice la columna `Pasajero`.

```python
# Descomentar si se produce un error de conexión relacionado con el certificado
#import ssl
#ssl._create_default_https_context = ssl._create_unverified_context

names = [
    'Pasajero',
    'Sobrevive',
    'Clase',
    'Nombre',
    'Sexo',
    'Edad',
    'Hermanos/Cónyuges',
    'Padres/Hijos',
    'Ticket',
    'Tarifa',
    'Número de Cabina',
    'Puerto de embarque'
]
df_titanic = pd.read_csv('https://www.dsi.uclm.es/personal/LuisDeLaOssa/recursos/titanic.csv', 
                         header=0, names=names, index_col="Pasajero")
df_titanic.head()
```

El método `DataFrame.to_csv()` escribe el contenido de un *DataFrame* en un archivo de texto. También acepta multitud de parámetros [**( + inf )**](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.to_csv.html), como por ejemplo:


* `header` o `index` para escribir los nombres de columnas o filas;
* `sep`, que indica la separación entre columnas;
* `na_rep` que especifica la  representación de los valores perdidos;
* `encoding`, con la codificación del archivo;
* etc. 


<div class="alert alert-block alert-info">

<i class="fa fa-info-circle" aria-hidden="true"></i>
Tanto `pd.read_csv()` como `DataFrame.to_csv()` admiten un parámetro, denominado `compression`, que permite leer o escribir, respectivamente, archivos comprimidos. La compresión reduce el espacio y la cantidad de datos que se leen o escriben, a cambio de trabajo adicional de compresión o descompresión. El efecto total sobre el tiempo depende del formato, el códec, los datos y el sistema de almacenamiento.

</div>

```python
compression = {"method": "zip", "archive_name": "titanic.csv"}

df_titanic.to_csv( "data/titanic.zip", sep=";", encoding="utf-8", compression=compression)
```

Es posible leer e inspeccionar archivos *Excel* con múltiples hojas mediante objetos de la clase `pd.ExcelFile` [**( + inf )**](https://pandas.pydata.org/docs/reference/api/pandas.ExcelFile.html).

```python
archivo_excel = pd.ExcelFile('data/alldata.xlsx')

hojas = archivo_excel.sheet_names
hojas
```

Se puede crear un `DataFrame` a partir de una hoja individual almacenada en un objeto `pd.ExcelFile` mediante `pd.read_excel()` o mediante el método `parse()`. 

```python
df_primera_hoja = pd.read_excel(archivo_excel, "Primera hoja")  # Modo más común
# df_primera_hoja = archivo_excel.parse(hojas[0])               # Alternativa

archivo_excel.close() # Lo cierra para no mantener recursos abiertos.

df_primera_hoja.head()
```

Para guardar varios `DataFrame` en un archivo *Excel*, o añadir hojas a un archivo existente, se ha de crear un objeto `pd.ExcelWriter` [**( + inf )**](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.ExcelWriter.html). En esta celda, por ejemplo, añade una copia del `DataFrame` `df_sequences` al archivo `data/alldata.xlsx`, en una hoja denominada `Segunda hoja`. 

```python
with pd.ExcelWriter("data/alldata.xlsx" , mode='a', engine="openpyxl", if_sheet_exists="replace") as writer:  
     df_sequences.to_excel(writer, sheet_name='Segunda hoja')
```

`if_sheet_exists="replace"` hace que, si la hoja existía ya en el archivo, se reemplace.  `if_sheet_exists="overlay"` hace que se reescriban las celdas correspondientes al `DataFrame` en la hoja, manteniendo el resto de celdas su valor.

### <font color="#5E5E5E"> Lectura y escritura de bases de datos </font>
*pandas* proporciona métodos para leer y escribir `DataFrames` en bases de datos *SQL*. En todos ellos es necesario proporcionar una conexión a la base de datos.

La función `pd.read_sql_table()` lee una tabla de una base de datos y la almacena en un `DataFrame`.

```python
 #%pip install "SQLAlchemy>=2.0.36"
```

```python
# sqlalchemy es un módulo que implementa las funciones
# para el acceso a bases de datos SQL.
from sqlalchemy import create_engine

# Crea una conexión a una base de datos sqlite
engine = create_engine('sqlite:///data/deportes.db')   # Abre un motor, pero no crea la conexión todavía.

# Lee la tabla
df_person = pd.read_sql_table("persona", engine, index_col="id")
df_person
```

La función `pd.read_sql_query()` permite hacer consultas sobre la base de datos y almacenar el resultado en un *DataFrame*.

```python
query = """
SELECT p.id, p.Apellido, p.Nombre, e.Equipo, e.Deporte
FROM persona AS p
JOIN equipo AS e ON p.id = e.id
"""

df_person_team = pd.read_sql_query(query, engine, index_col="id")
df_person_team
```

<div class="alert alert-block alert-info">
    
<i class="fa fa-info-circle" aria-hidden="true"></i> La función `pd.read_sql()` es una función de conveniencia: delega en `read_sql_table()` si recibe el nombre de una tabla, y en `read_sql_query()` si recibe una consulta SQL.
</div>

```python
#pd.read_sql("persona", engine, index_col="id")
pd.read_sql(query, engine, index_col="id")
```

Existen módulos de *python* que permiten trabajar directamente con algunos tipos de bases de datos específicos. Por ejemplo, `sqlite3` trabaja con bases de datos de tipo *sqlite*.

```python
# Bases de datos sqlite
import sqlite3 

conn = sqlite3.connect("data/deportes.db") # Crea la conexión

df_person_team = pd.read_sql_query(query, conn, index_col="id")
conn.close()

df_person_team
```

### <font color="#5E5E5E"> Archivos de gran tamaño </font>



Cuando aumenta el volumen de datos, el formato CSV puede resultar poco eficiente en espacio y tiempo de lectura. *Feather* y *Parquet* son formatos binarios de **almacenamiento por columnas** que permiten leer únicamente las columnas necesarias y guardar información sobre sus tipos de datos. Ambos admiten compresión [**( + inf )**](https://pandas.pydata.org/docs/user_guide/io.html).

*Feather* forma parte del ecosistema *Apache Arrow* y está orientado a la lectura, escritura e intercambio eficiente de datos entre herramientas de análisis. En *pandas*, se utiliza la función `pd.read_feather()` para leer estos archivos y el método `df.to_feather()` para escribir un `DataFrame` [**( + inf )**](https://pandas.pydata.org/docs/user_guide/io.html#feather).

*Parquet* se utiliza ampliamente en el almacenamiento y análisis de grandes conjuntos de datos. Permite organizar los datos en **particiones**, por ejemplo, separándolos por año o país, lo que puede facilitar la lectura de los subconjuntos de interés. En *pandas*, se utiliza `pd.read_parquet()` para la lectura y `df.to_parquet()` para la escritura [**( + inf )**](https://pandas.pydata.org/docs/user_guide/io.html#parquet).

La compresión reduce el espacio ocupado y puede acelerar la lectura al disminuir la cantidad de datos que se transfieren desde el almacenamiento. Su efecto sobre el rendimiento depende de los datos, del algoritmo y del equipo utilizado [**( + inf )**](https://arrow.apache.org/docs/python/feather.html#using-compression).

<div class="alert alert-block alert-info">

<i class="fa fa-info-circle" aria-hidden="true"></i>
Para trabajar con ambos formatos debe instalarse el paquete opcional <code>pyarrow</code>, que proporciona a <i>pandas</i> las herramientas necesarias para su lectura y escritura.

Al guardar con <code>df.to_feather()</code>, el <code>DataFrame</code> debe tener un índice predeterminado. Si tiene un índice personalizado que queremos conservar, podemos convertirlo en columna mediante <code>reset_index()</code> antes de guardarlo. Parquet permite conservar ese índice directamente <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.to_feather.html"><b>( + inf )</b></a>.

En otras asignaturas se tratarán con más detalle estos formatos y el trabajo con grandes conjuntos de datos.

### <font color="#5E5E5E"> Comprobación de tipos de datos</font>

El tipo de datos  determina los valores que puede contener, y las operaciones que pueden hacerse, con cada columna. Suele ser necesario explorar los tipos de las columnas inmediatamente después de leer un `DataFrame` para detectar incoherencias en este sentido. El método `DataFrame.info()` incluye esta información, que también puede ser accedida de manera específica, a través del atributo `DataFrame.dtypes`.

```python
df_sequences.dtypes
```

Según la salida del código anterior una de las columnas, `Publication year`, se codifica con números enteros; y otra, `Size (Mb)`, se codifica con números reales. 

Por otra parte, en pandas 3.0, las columnas cuyo contenido se reconoce como texto se infieren con el dtype `str`.

<div class="alert alert-block alert-info">

<i class="fa fa-info-circle" aria-hidden="true"></i>
En pandas 3.0, `str` y `string` tienen diferencias relevantes:

* `str` es el nuevo dtype predeterminado y representa los ausentes con `NaN`;
* `string` solicita el dtype anulable anterior, que utiliza `pd.NA`.
</div>

Las columnas `Kingdom`, `Class` y `Assembly status` se han inferido correctamente con el tipo `str`. Sin embargo, también se ha inferido este tipo de datos para las columnas `GC%`, `Number of genes` o `Number of proteins`, que expresan datos *numéricos*. 
*En este caso particular,* un vistazo al contenido de las últimas filas del `DataFrame` (con `DataFrame.tail()`) permite comprobar cual es el origen de este problema.

```python
df_sequences.tail()
```

Para estas columnas, el valor en algunas filas es el caracter `'-'`, que *no puede ser convertido a un número entero*. Debido a esto, la función `pd.read_csv()` determina que el contenido de la columna son cadenas de texto (`str`). Esto implica, por ejemplo, que el valor del número de genes de una de las filas con índice `Saccharomyces cerevisiae` se codifica como  `'155'` (`str`) y no `155` (`int`).

De esta inspección se deduce que, en el archivo `data/eukaryotes.tsv`, los datos perdidos se representan con el caracter `-`. Esta información se le puede proporcionar a `pd.read_csv()` mediante el parámetro `na_values`. 


```python
df_sequences = pd.read_csv("data/eukaryotes.tsv", sep="\t", index_col="Species", na_values="-")
df_sequences.head()
```

En este caso, al especificar cual es la representación de los valores nulos en el archivo, se infiere que el tipo de las columnas `GC%`, `Number of genes` y `Number of proteins` es numérico.

```python
df_sequences.info()
```

Un vistazo a los datos permite observar que las columnas `Number of genes` y `Number of proteins` deben contener valores enteros. Sin embargo, se han codificado como flotantes.  Existe una función denominada `DataFrame.convert_dtypes()` que hace una conversión, o más bien corrección, automática de tipos [**(+ inf)**](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.convert_dtypes.html) para estos casos.

```python
df_sequences = df_sequences.convert_dtypes()
```

```python
df_sequences.info()
```

<div class="alert alert-block alert-info">
    
<i class="fa fa-info-circle" aria-hidden="true"></i> `Number of genes` y `Number of proteins` se habían inferido como `float64` porque `np.NaN` es de ese tipo. Sin empargo, *pandas* implementa un tipo propio denominado `Int64` que es un *entero anulable*, y representa *not a number* con el valor `pd.NA` (al igual que lo hacen `Float64` o `string`).
</div>

```python
df_sequences.tail()
```

### <font color="#5E5E5E"> Inspección de los valores</font>


El método `DataFrame.describe()` / `Series.describe()` devuelve diversa información con respecto a las columnas. Esta información varía según el tipo de datos. Además, se puede especificar sobre qué columnas se aplica la función con los parámetros `include` \ `exclude`. La siguiente llamada incluye todas las columnas. Puede apreciarse que para las numéricas muestra unos datos, y para las no numéricas otros.

```python
#df_sequences.describe(include="all")
```

En el caso de los datos numéricos se devuelve, para cada columna, la media, la desviación, el mínimo y máximo valor, y los cuartiles. El argumento `percentiles` permite especificar qué cuantiles (en lugar de los cuartiles) se calculan. 

```python
df_sequences.describe(include="number")
#df_sequences.describe(include="number", percentiles=[0.2, 0.8])
```

Para las columnas no numéricas, `describe()` muestra el número de valores no ausentes (`count`), el número de valores distintos (`unique`), el valor más frecuente (`top`) y su frecuencia (`freq`). Si existe un empate, puede escoger uno de los valores más frecuentes.
