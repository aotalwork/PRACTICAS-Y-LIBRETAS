## <font color="#5E5E5E">1. Introducción</font>


En una libreta anterior se ha discretizado el valor de la columna `df_titanic['Tarifa']` mediante las funciones `pd.cut()` y `pd.qcut()`. Al mostrar las `Series` resultantes pueden observarse dos circunstancias.

```python
df_titanic = pd.read_csv('data/titanic_prep.csv', index_col=0)
df_titanic.head()
```

```python
df_titanic['Tarifa disc'] = pd.qcut(df_titanic['Tarifa'],3, labels=['Barata', 'Media', 'Cara'])
df_titanic['Tarifa disc'].head()
```

En primer lugar, el tipo de la serie es `category`, e incluye el número y conjunto de valores que puede tomar. Por otra parte, se establece una relación de orden entre ese conjunto de valores.

_pandas_ implementa un tipo de datos específico para el trabajo con datos categóricos, es decir, ___variables que pueden tomar un número finito y determinado de valores___. Aunque estos valores se visualizan como `strings`, _internamente se representan mediante enteros_, lo que permite un tratamiento eficiente. El uso de series/columnas con valores categóricos, aporta tres ventajas:

1. No se pueden asignar a los elementos valores que no estén recogidos en las categorías. Esto repercute en control sobre la integridad de los datos, y evita fallos.

```python
# df_titanic.loc[100, 'Tarifa disc']='Muy cara' # Produce un error
```

2. En segundo lugar, se define una relación de orden entre los valores. Se pueden utilizar operaciones de comparación para, por ejemplo seleccionar datos. 

```python
df_titanic[df_titanic['Tarifa disc']>'Media']
```

3. Por último, la representación categórica ocupa menos espacio en memoria que la representación equivalente en formato `string`.

```python
df_titanic['Tarifa str'] = df_titanic['Tarifa disc'].astype(str)
df_titanic.memory_usage(deep=True)
```

## <font color="#5E5E5E"><i class="fa fa-plus-circle" aria-hidden="true" style="color:#5E5E5E"></i> 2. El tipo de objetos `Categorical`</font>


Los objetos de tipo `pd.Categorical` permiten representar __una colección__ (no un `series`) de datos categóricos. Por defecto, si se contruye a partir de una lista de `strings`, considera tantas categorías como detecta en los datos, pero éstas se pueden especificar mediante el argumento `categories`.

```python
pd.Categorical(["primera","segunda","tercera","primera", "tercera", "especial"])
```

```python
# Solo considera tres categorías. 
pd.Categorical(["primera","segunda","tercera","primera", "tercera", "especial"],
                categories=["primera","segunda","especial"])
```

<div class="alert alert-block alert-info">

<i class="fa fa-info-circle" aria-hidden="true"></i> En una colección de valores categóricos, aquellos cuya categoría no ha sido especificada se codifican como `np.NaN`.
</div>

También es posible ___establecer una relación de orden___ entre las categorías.

```python
pd.Categorical(["primera","segunda","tercera","primera", "tercera", "especial"],
                categories=["primera","segunda","especial"], ordered=True)       # "tercera" no existe.
```

### <font color="#5E5E5E"> El tipo  `CategoricalDType` </font></font>

Un tipo de datos categórico se representa mediante `CategoricalDType`, y permite definir dos aspectos:

1. `categories`: El conjunto de valores categóricos.
2. `ordered`: Un valor booleano que representa si se establece un orden o no entre los valores categóricos. 



```python
from pandas.api.types import CategoricalDtype

tipo_calidades = CategoricalDtype(categories=["primera","segunda","especial"], ordered=True)
```

Es posible construir una secuencia de valores categóricos pasando como parámetro un objeto de tipo `CategoricalDtype` a través del parámetro `dtype`.

```python
# Esto es equivalente a la construcción anterior.
pd.Categorical(["primera","segunda","tercera","primera", "tercera", "especial"], dtype=tipo_calidades)
```

Tanto las categorías como el orden utilizados en un objeto `Categorical` pueden accederse a través de `Categorical.categories` y `Categorical.ordered`.

```python
calidades = pd.Categorical(["primera","segunda","tercera","primera", "tercera", "especial"], dtype=tipo_calidades)
print(calidades.categories)
print(calidades.ordered)
```

<div class="alert alert-block alert-warning">

<i class="fa fa-exclamation-circle" aria-hidden="true"></i>
En este punto es importante distinguir entre `CategoricalDtype` que permite definir los valores que puede tomar una categoría, y el objeto `Cagorical` que contiene una secuencia de valores del tipo `CategoricalDtype`.
</div>

### <font color="#5E5E5E"> Renombrado de categorías</font>

Es posible renombrar las distintas categorías consideradas en un objeto `Categorical`  mediante el método `Categorical.rename_categories()`. Éste puede tomar como argumentos una lista, o un diccionario. La operación redefine tanto el tipo de datos (etiquetas de las categorías) como los valores almacenados, y produce un nuevo objeto `Categorical`.

```python
calidades.rename_categories(['1ª','2ª','Extra'])   # Reemplaza los valores
```

```python
calidades.rename_categories({'segunda':'2ª',
                             'especial':'Extra',
                             'primera':'1ª'})                      # Los vuelve a reemplazar. 
```

### <font color="#5E5E5E"> Reemplazo de categorías</font>

El método `Categorical.set_categories()` permite reestablecer las categorías. Su aplicación devuelve un nuevo objeto de tipo `Categorical`. Si alguna categoría desaparece, los valores correspondientes pasan a ser `np.NaN`.

```python
calidades.set_categories(['primera','segunda'])
```

### <font color="#5E5E5E"> Adición de categorías</font>

Es posible añadir nuevas categorías mediante el método `Categorical.add_categories()`.

```python
calidades = calidades.add_categories(['tercera','superior'])
calidades
```

<div class="alert alert-block alert-warning">

<i class="fa fa-exclamation-circle" aria-hidden="true"></i> Los valores `NaN` siguen siendo `NaN` tras aplicar las dos funciones anteriores.
</div>

### <font color="#5E5E5E"> Eliminación de categorías</font>

El método `Categorical.remove_categories()` permite eliminar categorías. 

```python
calidades = calidades.remove_categories(['superior'])
calidades
```

Además, existe un método, denominado `Categorical.remove_unused_categories()` que permite eliminar las categorías que no han sido utilizadas.

```python
calidades = calidades.remove_unused_categories() # Elimina la categoría superextra
calidades
```

### <font color="#5E5E5E"> Gestión del orden </font>

Los métodos `Categorical.as_ordered()` y `Categorical.as_unordered()` permiten, respectivamente, establecer o eliminar el orden establecido entre las etiquetas.

```python
calidades.as_unordered()
```

El método `Categorical.reorder_categories()` permite cambiar el orden de las categorías. 

```python
calidades.reorder_categories(["segunda", "especial", "primera"], ordered=True)
```

### <font color="#5E5E5E"> Unión de conjuntos de datos categóricos </font>

`union_categoricals()` concatena los valores de varios objetos categóricos y construye un conjunto de categorías que incluye las etiquetas de todos ellos. En este ejemplo ambos objetos son no ordenados y sus categorías no se solapan.

```python
from pandas.api.types import union_categoricals

calidades1 = pd.Categorical(["Limón", "Naranja", "Naranja", "Limón"])
calidades2 = pd.Categorical(["Melón", "Sandía", "Sandía", "Melón"])

union_categoricals([calidades1, calidades2])
```

## <font color="#5E5E5E">3. Series con valores categóricos </font>

Se puede crear un objeto `Series` de valores categóricos directamente especificando (a través de `dtype`) el tipo `category`. En ellos, los valores (`Series.values`) son un objeto de tipo `Categorical`. 

```python
# Crea una serie con valores de tipo category
pd.Series(["primera","segunda","tercera","primera", "tercera", "especial"], dtype="category")
```

Si no se especifica el tipo, las etiquetas se consideran `str`, pero se puede convertir mediante `Series.astype()`.

```python
# Crea una serie de valores de tipo String
pd.Series(["primera","segunda","tercera","primera", "tercera", "especial"])
```

```python
# Convierte la serie a valores de tipo category
calidades.astype("category")
```

También es posible crear la serie pasando como tipo un objeto  `CategoricalDtype`.

```python
tipo_calidades = CategoricalDtype(categories=["primera", "segunda", "tercera", "especial"], ordered=True)
calidades = pd.Series(["primera","segunda","tercera","primera", "tercera", "especial"], dtype=tipo_calidades)
calidades
```

Las funciones de agregación se adaptan al trabajo con categorías.

```python
calidades.min()
calidades.mode()
```

```python
calidades.value_counts()
```

También es posible utilizar los accesores `str` y `dt` (se verá después) si las etiquetas son compatibles.

```python
calidades.str.upper()
```

### <font color="#5E5E5E"> Manipulación de las categorías en Series</font>

`Series.cat` proporciona acceso a todas las funciones descritas para `Categorical`.


<div class="alert alert-block alert-info">

<i class="fa fa-info-circle" aria-hidden="true"></i> Las funciones que proporciona el objeto `Categorical` que contiene los valores del `Series` se __deben acceder__ a través de la propiedad `Series.cat`. 
</div>

```python
print(calidades.cat.categories)
print(calidades.cat.ordered)
#print(calidades.cat)
#print(calidades.values)
#print(calidades.cat.codes)
```

#### <font color="#5E5E5E"> Renombrado de categorías</font>

```python
calidades = calidades.cat.rename_categories(['1ª','2ª','3ª','Extra'])
# calidades.cat.categories = ['1ª','2ª','3º','Extra']     # Equivalente
calidades
```

```python
# También puede tomar un diccionario
calidades = calidades.cat.rename_categories({'1ª':'primera',
                                             '2ª':'segunda',
                                             '3ª':'tercera',
                                             'Extra':'especial'})
calidades
```

#### <font color="#5E5E5E"> Reemplazo de categorías</font>

```python
calidades = calidades.cat.set_categories(["primera", "segunda", "tercera", "extra"], ordered=True)
calidades # Extra no existía
```

#### <font color="#5E5E5E"> Adición y eliminación </font>

```python
# Añade una categoría
calidades = calidades.cat.add_categories(["especial"])
calidades
```

```python
# Elimina la categoría
calidades = calidades.cat.remove_categories(["especial"])
print(calidades)
```

#### <font color="#5E5E5E"> Eliminación de categorías no utilizadas </font>

```python
calidades = calidades.cat.remove_unused_categories()
calidades
```

#### <font color="#5E5E5E"> Reordenación </font>

```python
# Elimina el orden
calidades = calidades.cat.as_unordered()
calidades
```

```python
# Vuelve a usar el orden (no determina cual)
calidades.cat.as_ordered()
```

```python
# Determina el orden
calidades.cat.reorder_categories(["segunda","primera","tercera"], ordered=True)
```

### <font color="#5E5E5E"> Ordenación según categorías</font>

Es posible ordenar los *Series* en función de las etiquetas mediante `Series.sort_values()`.

```python
tipo_calidades = CategoricalDtype(categories=["primera","segunda","especial"], ordered=True)
calidades = pd.Series(["primera","segunda","tercera","primera", "tercera", "especial"], dtype=tipo_calidades)
```

```python
calidades.sort_values(inplace=False)
```

### <font color="#5E5E5E"> Operaciones con Series de valores categóricos.</font>
<br>

Es posible llevar a cabo operaciones de comparación entre `Series` de datos categóricos **cuando utilizan la misma categoría**.

```python
tipo_calidades = CategoricalDtype(categories=["segunda","primera","especial"], ordered=True)

calidades1 = pd.Series(["primera","segunda","tercera","especial", "tercera", "especial"], dtype=tipo_calidades)
calidades2 = pd.Series(["segunda","segunda","segunda","primera", "especial", "especial"], dtype=tipo_calidades)

print(calidades1>calidades2)
#print(calidades1==calidades2)
#print(calidades1>"primera")
```

## <font color="#5E5E5E"><i class="fa fa-plus-circle" aria-hidden="true" style="color:#5E5E5E"></i> 4. Indexación con valores categóricos.</font>


También es posible utilizar valores categóricos en el índice de las `Series`.


```python
niveles = pd.Categorical(
    ["Bajo", "Muy alto", "Medio", "Alto"],
    categories=["Bajo", "Medio", "Alto", "Muy alto"],
    ordered=True
)

df_clientes = pd.DataFrame({
    "satisfaccion": niveles,
    "numero_clientes": [15, 68, 31, 42]
}).set_index("satisfaccion")

df_clientes
```

```python
# Índice y uso del índice
print(type(df_clientes.index),'\n')
print(df_clientes.loc["Alto"],'\n')
```

 La indexación por categorías permite también cierto grado de _slicing_. Las etiquetas utilizadas deben ser únicas.

```python
# Slicing
df_clientes.sort_index()["Bajo":"Alto"]
```
