## <font color="#5E5E5E">1. Introducción</font>


Otra funcionalidad interesante de _pandas_ es la capacidad de unión de conjuntos de datos. La documentación oficial de la librería ilustra con ejemplos la mayoría de casos de uso [ **( + inf )**](https://pandas.pydata.org/pandas-docs/stable/user_guide/merging.html). Básicamente, existen dos formas de combinar conjuntos de datos:

* **concatenación**, que simplemente se define a nivel estructural. Combina elementos en un eje, y alinea las etiquetas en el otro.
* **fusión**, que considera relaciones entre los datos, etableciendo correspondencias por claves. 

<div class="alert alert-block alert-info">

<i class="fa fa-info-circle" aria-hidden="true"></i>
La fusión es un proceso que repercute en la organización de los conjuntos de datos, y existen algunas consideraciones en este sentido. En este tutorial se abordan las herramientas que proporciona *pandas* para la fusión de `DataFrame`. 
</div>

Para ilustrar los ejemplos de esta libreta, se utilizarán estos tres *DataFrame* sintéticos de tamaño reducido.

```python
import pandas as pd

# Posición - País
df_PP = pd.DataFrame([
    ('Erling Haaland', 'Delantero', 'Noruega'),
    ('Virgil van Dijk', 'Defensa', 'Países Bajos'),
    ('Pedri', 'Centrocampista', 'España'),
    ('Kylian Mbappé', 'Delantero', 'Francia'),
    ('Thibaut Courtois', 'Portero', 'Bélgica')
], columns=['Nombre','Posición','País']).set_index('Nombre')


# Posición - País - Equipo
df_PPE = pd.DataFrame([
    ('Lionel Messi', 'Delantero', 'Argentina', 'Inter Miami'),
    ('Luka Modrić', 'Centrocampista', 'Croacia', 'Real Madrid'),
    ('Frenkie de Jong', 'Centrocampista', 'Países Bajos', 'FC Barcelona'),
    ('Karim Benzema', 'Delantero', 'Francia', 'Al-Ittihad'),
    ('Kevin De Bruyne', 'Centrocampista', 'Bélgica', 'Manchester City'),
    ('Erling Haaland', 'Delantero', 'Noruega', 'Manchester City')
], columns=['Nombre','Posición','País','Equipo']).set_index('Nombre')


# Equipo - País
df_EP = pd.DataFrame([
    ('Lionel Messi','Inter Miami','Estados Unidos'),
    ('Kylian Mbappé','Paris Saint-Germain','Francia'),
    ('Kylian Mbappé','Real Madrid','España'),
    ('Karim Benzema','Al-Ittihad','Arabia Saudí'),
    ('Luka Modrić','Real Madrid','España'),
    ('Pedri','FC Barcelona','España'),
    ('Erling Haaland','Manchester City','Inglaterra'),
    ('Kevin De Bruyne','Manchester City','Inglaterra'),
    ('Virgil van Dijk','Liverpool','Inglaterra')
], columns=['Nombre','Equipo','País']).set_index('Nombre')

print("DF_PP:\n", df_PP, "\n")
print("DF_PPE:\n", df_PPE, "\n")
print("DF_EP:\n", df_EP, "\n")
```

## <font color="#5E5E5E">2. Concatenación de *Series* y *DataFrames*</font>



### <font color="#5E5E5E">  `pd.concat()`  </font>

Esta función implementa la concatenación de `Series` o `DataFrame`. Puede hacerse tanto a nivel de filas como de columnas, y opera sobre **una colección de** estructuras. Toma varios parámetros [**( + inf )**](https://pandas.pydata.org/pandas-docs/stable/generated/pandas.concat.html). Los más importantes, además de la lista de `Series` / `DataFrame` que se han de concatenar, son:

* `axis`, que determina el eje a lo largo del cual se concatenan los datos, y puede tomar los valores `0/index` y `1/columns`;
* `join`, que especifica si se considera la unión (`outer`) o la intersección (`inner`) de elementos en el otro eje (se podrían hacer `left` y `right` procesando el resultado con `DataFrame.reindex()`. 
* `keys`, que es un vector de claves. Si se utiliza, crea un índice de dos niveles, y utiliza estas claves en el primer nivel para marcar el `DataFrame` original en el resultante. 

La siguiente celda muestra la concatenación de una colección  `Series`. Puede apreciarse que, puesto que el nombre del índice es común en ambas, se preserva.

```python
pd.concat([df_PP['País'], df_PPE['País']]) # axis=0 por defecto
```

El indice de la serie resultante tiene valores repetidos.  El parámetro `verify_integrity=True` hace que `pd.concat` lance una excepción si ocurre esto.

Podrían concatenarse las `Series` con `axis=1` (aunque no tiene mucho sentido).

```python
#pd.concat([df_PP['País'], df_PPE['País']], axis=1)
```

En el caso de los `DataFrame`, la llamada a `pd.concat(lista)` crea un nuevo `DataFrame` con la concatenación de las filas/columnas de los `DataFrame` incluidos en `lista`. El parámetro `join` determina qué conjunto de índices **del eje que no se concatena** se incluye en el `DataFrame` resultado. Por defecto es `outer`, es decir, utiliza la unión. Cuando se unen `DataFrame` con distintas columnas, y se usa `join=outer`, los valores no existentes se fijan a `NaN` en el nuevo `DataFrame`. 

```python
pd.concat([df_PP, df_PPE])   # Por defecto, join="outer" y axis=0
# pd.concat([df_PP, df_PPE], join="outer", axis=0)  
```

Mediante `join=inner` se indica que se incluyan solamente las columnas que aparecen en ambos conjuntos.

```python
pd.concat([df_PP, df_PPE], join="inner", axis=0)
```

Mediante el parámetro `keys` se añade al índice una clave que permite identificar el origen de los datos. El resultado es un `DataFrame` con un índice de dos niveles.

```python
pd.concat([df_PP, df_PPE], join="inner", axis=0, keys=["JUG", "JUG EQUIPO"])
```

Con `axis=1` se concatenan las columnas. En este ejemplo, se indica que solamente se consideren aquellas filas cuyo índice aparece en ambos `DataFrame` mediante `join=inner`.

```python
pd.concat([df_PP, df_PPE], axis=1, join='inner')
```

Si se especifica `join=outer` se incluyen todas las filas. 

```python
pd.concat([df_PP, df_PPE], axis=1, join='outer')
```

<div class="alert alert-block alert-warning">

<i class="fa fa-exclamation-circle" aria-hidden="true"></i>
Estas dos operaciones producen error si alguno de los *DataFrame* implicados tiene índices repetidos.
</div>

```python
#pd.concat([df_PP, df_EP], axis=1, join='outer') # Produce un error por Mbappé
```

<div class="alert alert-block alert-warning">

<i class="fa fa-exclamation-circle" aria-hidden="true"></i>
Hay que tener en cuenta, por cuestiones de eficiencia y memoria, que `pd.concat()` hace una copia de los datos. Por ejemplo, no es buena idea utilizarlo para añadir filas a un `DataFrame` existente.
</div>

## <font color="#5E5E5E">3. Fusión de *DataFrames*</font>




### <font color="#5E5E5E">   `pd.merge()`/`DataFrame.merge()`  </font>

La función/método `pd.merge(left, right)` / `DataFrame.merge(right)` (son equivalentes) permite unir **dos** `DataFrame` mediante una o varias columnas denominadas *claves de unión*, de modo que se establece una **correspondencia entre las filas** de ambos `DataFrame` en función del valor de esas columnas (es similar a las uniones de *SQL*). El resultado es un nuevo `DataFrame` con la unión de las columnas de los `DataFrame` originales, y en el que el conjunto de filas depende de cómo se opere (parámetro `how`) con los conjuntos de claves de unión de ambos `DataFrame`:

* **Unión** (`how='outer'`): el `DataFrame` resultante contiene las filas correspondientes a todas las claves de unión presentes en cualquiera de los dos `DataFrame`.
* **Intersección** (`how='inner'`): el `DataFrame` resultante contiene solo las filas cuyas claves de unión aparecen en los dos `DataFrame` originales.
* **Izquierda** (`how='left'`): el `DataFrame` resultante contiene las filas correspondientes a todas las claves de unión del `DataFrame` izquierdo, tengan o no correspondencia en el derecho.
* **Derecha** (`how='right'`): el `DataFrame` resultante contiene las filas correspondientes a todas las claves de unión del `DataFrame` derecho, tengan o no correspondencia en el izquierdo.

* **Anti-join izquierdo** (`how='left_anti'`): el `DataFrame` resultante contiene únicamente las filas del `DataFrame` izquierdo cuyas claves de unión **no aparecen** en el `DataFrame` derecho.
* **Anti-join derecho** (`how='right_anti'`): el `DataFrame` resultante contiene únicamente las filas del `DataFrame` derecho cuyas claves de unión **no aparecen** en el `DataFrame` izquierdo.


Acepta varios parámetros [**( + inf )**](https://pandas.pydata.org/pandas-docs/stable/generated/pandas.DataFrame.merge.html) que rigen la unión. Los más importantes son:

* `left`, `right`: que son argumentos posicionales que referencian a los dos `DataFrame` que son unidos. 
* `left_index`, `right_index`: que determinan los índices respectivos se usan como claves de unión.
* `on`, `left_on`, `right_on`: que determinan qué columnas (si no se usan índices) son utilizadas como claves de unión. `on` se utiliza cuando las columnas aparecen en ambos `DataFrame`. También se puede especificar el nombre del índice.

Además, admite otros parámetros de utilidad a la hora de presentar el conjunto de datos resultante de la unión.

* `suffixes`. Es una lista de `Strings` (dos). Cuando existen columnas comunes en ambos `DataFrame`, y no son utilizadas como clave de unión, permite identificarlas en el `DataFrame` resultante. Para ello, añade cada `string` al nombre de la columna correspondiente según incluya los valores de uno u otro `DataFrame`.  
* `indicator`. Añade una columna, denominada `_merge` con información sobre el origen de cada fila (un `DataFrame` concreto o los dos).
* `validate`. Es un `string` que permite determinar si se cumple una determinada relación entre las claves de unión. Puede tomar los valores `1:1`, `1:m`, `m:1` y `m:m`. Si no se cumple la relación, se genera una excepción. **Es muy importante** para controlar el resultado de la unión.

En la siguiente celda se lleva a cabo la unión entre los dos `DataFrame` definidos anteriormente en función del índice, y considerando la intersección de todas las filas. Como la columna *País* aparece en ambos `DataFrame`, se añade también un sufijo para determinar la correspondencia en el `DataFrame` resultante.

```python
pd.merge(df_PP, df_EP, how='inner', left_index=True, right_index=True, suffixes=['_jug','_equ'], validate='1:m')  
```

Como en el `df_EP` hay dos filas cuyo índice es `Kylian Mbappé`, la salida contiene la combinación de la fila del `DataFrame` `df_PP` con cada una de ellas. Por otra parte, la relación es `1:m`.

<div class="alert alert-block alert-warning">

<i class="fa fa-exclamation-circle" aria-hidden="true"></i>
Si un valor de la columna de unión aparece varias veces en el `DataFrame` de la izquierda, y varias en el de la derecha, en el resultante aparecen todas las combinaciones (producto cartesiano).
</div>

Es también correcto utilizar el nombre del índice (como si fuese una columna). En este caso, por ejemplo, se especifica `how='outer'`, por lo que se muestra la unión de las filas, rellenando con el valor `NaN` todos los valores inexistentes.

```python
pd.merge(df_PP,df_EP, how='outer', on='Nombre', suffixes=['_jug','_equ'])  
```

En este caso, `left_on` y `right_on` toman el valor del nombre del índice, `Nombre`.

```python
pd.merge(df_PP, df_EP, how='outer', left_on='Nombre', right_on='Nombre', suffixes=['_jug','_equ'])  # Equivalente
```

Cuando se especifica `how='left'` se incluyen las filas del `DataFrame` de la izquerda. Es posible crear una columna con la procedencia de cada fila (para utilizarla posteriomente) mediante `indicator=True`.

```python
pd.merge(df_PP,df_EP, how='left', on='Nombre', suffixes=['_jug','_equ'], indicator=True)  
```

`indicator` también admite un nombre de columna.

```python
pd.merge(df_PP,df_EP, how='left', on='Nombre', suffixes=['_jug','_equ'], indicator='Origen')  
```

Aunque es algo distinto de la funcionalidad de `pd.merge()` porque no utiliza clave de unión, `pd.merge()` permite hacer un producto cartesiano de las filas mediante `how='cross'`.

```python
pd.merge(df_PP,df_EP, how='cross')  # El resultado no tiene sentido, pero así se ve el efecto.
```

<div class="alert alert-block alert-warning">

<i class="fa fa-exclamation-circle" aria-hidden="true"></i>
Cuando la clave de unión son categorías, hay que controlar si se ha modificado la categoría resultante. 
</div>

### <font color="#5E5E5E"> `DataFrame.join()`  </font>

Es similar a `pd.merge()`/`DataFrame.merge()` , aunque permite unir las columnas de **varios** `DataFrame` y utiliza solo algunos parámetros. Por defecto utiliza los índices como clave de unión y el conjunto de elementos del `DataFrame` de la izquierda, es decir, `how=left` [**( + inf)**](https://pandas.pydata.org/pandas-docs/stable/generated/pandas.DataFrame.join.html).

```python
df_PP.join(df_EP, lsuffix='_jug', rsuffix='_equ') # En este caso se une un dataframe, pero podrían ser más. 
#df_PP.join([df_EP])                               # Si en la derecha hay una colección, no acepta sufijos
```

<div class="alert alert-block alert-info">

<i class="fa fa-info-circle" aria-hidden="true"></i>
Existe otro método, menos utilizado pero muy útil, denominado `merge_asof()`, que relaciona cada fila de la izquierda con una clave cercana de la derecha. Requiere datos ordenados y, por defecto, toma la última clave no posterior. Si se especifica `direction='nearest'` elige la más cercana. [**(+ inf)**](https://pandas.pydata.org/docs/reference/api/pandas.merge_asof.html).
    
Otro método, algo más complejo, es `pd.merge_ordered()`. Une también `DataFrame` ordenados, como series temporales, y permite rellenar valores perdidos mediante interpolación  [**(+ inf)**](https://pandas.pydata.org/docs/reference/api/pandas.merge_ordered.html#pandas.merge_ordered).
