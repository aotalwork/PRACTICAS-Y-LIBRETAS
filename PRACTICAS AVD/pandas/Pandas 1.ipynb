## <font color="#5E5E5E"> 2. Ejemplo</font>
<br>

En esta primera libreta se ilustra, mediante un breve *ejemplo preliminar*, el tipo de procesamiento de datos que se puede hacer con *pandas*. 

El archivo `'./data/fifa19.csv'` contiene un conjunto de datos sobre jugadores en el videojuego *FIFA 2019*, y fue obtenido mediante *scraping* de la web [(sofifa.com)](https://sofifa.com/players). Contiene información sobre 86 características para 18207 jugadores.
El siguiente código lee el archivo, lo almacena en un objeto de la clase `DataFrame`, y muestra su cabecera. 

```python
# Lee el CSV usando la primera columna como índice provisional
# y la sustituye por el identificador de cada jugador.
df_fifa = pd.read_csv('./data/fifa19.csv', index_col=0).set_index('ID')

# Muestra la cabecera (5 primeras filas)
df_fifa.head()
```

En esta celda se seleccionan varias columnas de las 100 primeras filas (100 primeros jugadores).

```python
# Lista con los nombres de las columnas de interés
sel_columns = ['Name', 'Age', 'Nationality', 'Overall', 'Club', 'Value', 'Wage', 'Position', 'Joined', 'Height', 'Weight', 'Release Clause']

# Selecciona filas y columnas
df_fifa = df_fifa.head(100)[sel_columns] 

# Muestra la cabecera
df_fifa.head(10)
```

Puede observarse que la altura de los jugadores se expresa mediante el sistema imperial (británico), y está representada con una cadena de caracteres (`str`).  En esta celda, se convierten las unidades de la columna `Height` al sistema métrico decimal (metros), y se representan mediante un _valor numérico_. 

<div class="alert alert-block alert-info">
    
<i class="fa fa-info-circle" aria-hidden="true"></i>
Por claridad en el código del tutorial, se evitará hacer control de excepciones a menos que sea estrictamente necesario.
</div>

```python
feet_to_cm = 30.48 # Pies a centímetros
inch_to_cm = 2.54  # Pulgadas a centímetros

# Toma un String con formato "pies'pulgadas" y devuelve un número con los metros
def imp_to_metric(imp_height):
    feet, inch = imp_height.split("'")                             # Separa pies y pulgadas 
    return (float(feet)*feet_to_cm + float(inch)*inch_to_cm)/100   # Convierte a float y calcula los metros
```

```python
# Convierte toda la columna utilizando la función `imp_to_metric`.
df_fifa['Height'] = df_fifa['Height'].apply(imp_to_metric)    

# Lo muestra
df_fifa.head()
```

En esta celda se transforman peso, salario y valor,  de manera más compacta, mediante funciones _lambda_.

```python
df_fifa['Weight'] = df_fifa['Weight'].apply(lambda w: float(w[:-3])*0.453592) # Libras a kg.
df_fifa['Value'] = df_fifa['Value'].apply(lambda v: float(v[1:-1]))           # Convierte en numérico.
df_fifa['Wage'] = df_fifa['Wage'].apply(lambda w: int(w[1:-1])*1000)          # Convierte a numérica la ganancia semanal en euros

df_fifa.head()
```

Al igual que en las bases de datos relacionales, se puede hacer _selección condicional_ de filas de un `DataFrame`. Por ejemplo, se pueden mostrar los jugadores cuyo valor está por encima de 100 millones de euros.

```python
df_fifa[df_fifa['Value']>100]
```

*pandas* permite obtener *directamente* estadísticos descriptivos simples. En la siguiente celda, se describe la información que contienen las columnas cuyo tipo es numérico.

```python
df_fifa.describe(include="number", percentiles=[0.25, 0.5, 0.75])
```

También podemos contar cuántos de los 100 jugadores seleccionados pertenecen a cada club. El resultado se ordena de mayor a menor frecuencia.

```python
df_fifa['Club'].value_counts()
```

Por último, en esta celda se muestra el número de jugadores de cada club que están entre los 100 seleccionados, así como el valor de su jugador más caro. Cuando las operaciones tienen cierta complejidad, pueden encerrarse entre paréntesis y expresarse en varias líneas para facilitar la lectura e interpretación del código.

```python
(df_fifa.groupby('Club')['Value']                                    # Agrupa los jugadores por club
        .agg(Jugadores='size', Maximo='max')                         # Calcula y almacena los resultados en las columnas Jugadores y Maximo
        .rename(columns = {'Jugadores': 'Numero de jugadores',       # Renombra las columnas
                 'Maximo': 'Valor máximo'})
)
```

# <font color="#5E5E5E" size=5>3. `DataFrame` y `Series` </font>


*pandas* trabaja con dos tipos de objetos *principales*, los `DataFrame` y las `Series`. 

Un `DataFrame` es una estructura de datos *bidimensional* organizada en filas y columnas. Cada eje tiene sus etiquetas: `index` identifica las filas y `columns`, las columnas. Cada columna tiene un tipo de datos (`dtype`).

La siguiente celda muestra el `DataFrame` `df_fifa`, que contiene un subconjunto de los datos del archivo `./data/fifa19.csv`, leídos y procesados en la sección anterior. En este ejemplo hay columnas de texto, como `Nationality`, y numéricas, como `Age` o `Height`.

```python
df_fifa
```

```python
df_fifa.dtypes
```

La visualización de un `DataFrame` permite distinguir, en negrita, **los índices**. En el caso del `DataFrame` `df_fifa`, el índice de filas, o ***índice*** simplemente, se denomina `ID`, y contiene un número de identificación de cada jugador. Puede apreciarse que las etiquetas de índice, a pesar de ser numéricas, **son independientes la posición** de las filas. Por ejemplo, la primera fila tiene la etiqueta *158023*, aunque su posición es *0*. El índice de ***columnas***, por otra parte, suele estar formado por cadenas de caracteres que describen el contenido de las mismas. 

<div class="alert alert-block alert-info">
    
<i class="fa fa-info-circle" aria-hidden="true"></i>
Los nombres de las columnas han de ser descriptivos, pero también ha de tenerse en cuenta que se utilizarán en el código, por lo que es conveniente que no sean excesivamente largos.
</div>

Cada columna de un `DataFrame` puede ser accedida y tratada de manera individual como un objeto de tipo `Series`, que es una estructura de datos *unidimensional*, en la que los valores tienen una posición, pero que cuenta con un índice a través del cual se accede a sus elementos.

```python
df_fifa['Nationality']
```

<div class="alert alert-block alert-info">
    
<i class="fa fa-info-circle" aria-hidden="true"></i>
El índice del objeto `Series` correspondiente a la columna es el índice del `DataFrame`. El nombre, es el nombre de la columna.
</div>
