### <font color="#5E5E5E">7.2 Operaciones vectorizadas sobre *Series* de *Strings*: `Series.str`</font>

### <font size=3 color="#5E5E5E"> <i class="fa fa-pencil-square-o" aria-hidden="true" style="color:#5E5E5E"></i> Ejercicio 24</font> 

Transformar los valores de la columna `Marital_Status` del *DataFrame* `df_marketing` a minúscula. Asignar la *Serie* resultante a la propia columna y mostrar la cabecera de `df_marketing`.

```python
df_marketing["Marital_Status"] = df_marketing["Marital_Status"].apply(lambda x: x.upper() if isinstance(x, str) else x)
df_marketing.head()
```

### <font size=3 color="#5E5E5E"> <i class="fa fa-pencil-square-o" aria-hidden="true" style="color:#5E5E5E"></i> Ejercicio 25</font> 

Utilizar `Series.str.extract()` para descomponer la fecha almacenada en la columna `Dt_Customer` del *DataFrame* `df_marketing` en tres columnas; `Day`, `Month` y `Year`.

```python
regex = (r'(?P<Day>\d+)-(?P<Month>\d+)-(?P<Year>\d+)')
df_marketing["Dt_Customer"].str.extract(regex, expand=True)

```

### <font size=3 color="#5E5E5E"> <i class="fa fa-pencil-square-o" aria-hidden="true" style="color:#5E5E5E"></i> Ejercicio 26</font> 

Repetir el ejercicio anterior, pero convirtiendo las columnas resultantes a enteros con `DataFrame.astype('int64')`, y asignando el resultado a nuevas columnas del *DataFrame* `df_marketing` denominadas `Day`, `Month` y `Year`.

```python
df_marketing[["Day", "Month", "Year"]] = df_marketing["Dt_Customer"].str.extract(regex, expand=True).astype({"Day": "int64", "Month": "int64", "Year": "int64"})
```

### <font color="#5E5E5E">7.3 Operaciones de transformación sobre *Series*</font>

### <font size=3 color="#5E5E5E"> <i class="fa fa-pencil-square-o" aria-hidden="true" style="color:#5E5E5E"></i> Ejercicio 27</font> 

Crear una nueva columna denominada `Season` en el *DataFrame* `df_marketing`, y a partir de la columna `Month`, de modo que el valor de `Season` sea: `Winter` para los meses 1 a 3; `Spring` para los meses 4 a 6; `Summer` para los meses 5 a 9; y `Autumn` para los meses 10 a 12. Utilizar un diccionario para definir las transformarciones.

```python
season_dict = {1: 'Winter', 2: 'Winter', 3:'Winter', 
               4: 'Spring', 5: 'Spring', 6: 'Spring', 
               7: 'Summer', 8: 'Summer', 9: 'Summer',
               10: 'Autumn',11:'Autumn', 12: 'Autumn'}

df_marketing["Season"] = df_marketing["Month"].map(season_dict)
df_marketing["Season"]
```

### <font size=3 color="#5E5E5E"> <i class="fa fa-pencil-square-o" aria-hidden="true" style="color:#5E5E5E"></i> Ejercicio 28</font> 

Repetir el ejercicio anterior, pero utilizando `Series.apply()`.

```python
def get_season(month):
    if month<4:
        return 'Winter'
    elif month<7:
        return 'Spring'
    elif month<10:
        return 'Summer'
    else:
        return 'Winter'
    

df_marketing["Season"] = df_marketing["Month"].apply(lambda x: get_season(x))
df_marketing["Season"]
```

### <font size=3 color="#5E5E5E"> <i class="fa fa-pencil-square-o" aria-hidden="true" style="color:#5E5E5E"></i> Ejercicio 29</font> 


A partir de la columna `Month` del *DataFrame* `df_marketing`, generar un nuevo *DataFrame* con dos columnas, de modo que la primera de ella contenga el nombre del mes correspondiente, y la segunda la estación. Utilizar una función *lambda* para acceder al nombre del mes.  

```python
months = ['January', 'February','March',
          'April','May','June',
          'July','August','September',
          'October','November','December']

df_marketing_nuevo = df_marketing[["Month", "Season"]]

df_marketing_nuevo["Month"] = df_marketing_nuevo["Month"].apply(lambda x: months[x-1])
df_marketing_nuevo
```

### <font size=3 color="#5E5E5E"> <i class="fa fa-pencil-square-o" aria-hidden="true" style="color:#5E5E5E"></i> Ejercicio 31</font>

### <font size=3 color="#5E5E5E"> <i class="fa fa-pencil-square-o" aria-hidden="true" style="color:#5E5E5E"></i> Ejercicio 32</font> 

Crear una columna denominada `MntTotal` que contenga la suma de `MntWines`, `MntFruits`, `MntMeatProducts`, `MntFishProducts`, `MntSweetProducts` y `MntGoldProds`. Utilizar operaciones vectorizadas.

### <font size=3 color="#5E5E5E"> <i class="fa fa-pencil-square-o" aria-hidden="true" style="color:#5E5E5E"></i> Ejercicio 39</font> 


Obtener, con una sola orden, la media,  desviación y suma de la columna `Income`. Indexar los elementos resultantes utilizando las etiquetas `Media`, `Desviación` y `Suma` respectivamente.

```python
media, desviacion_estandar, suma = df_marketing["Income"].mean(), df_marketing["Income"].std(), df_marketing["Income"].sum()
print("Media:", media)
print("Desviación estándar:", desviacion_estandar)
print("Suma:", suma)
```

### <font size=3 color="#5E5E5E"> <i class="fa fa-pencil-square-o" aria-hidden="true" style="color:#5E5E5E"></i> Ejercicio 40</font> 

Utilizando una sola orden, obtener la media y desviación para todas las columnas en `mnt_columns`.


```python
mnt_colums = ['MntWines', 'MntFruits', 'MntMeatProducts', 'MntFishProducts', 'MntSweetProducts', 'MntGoldProds']

media, desviacion_estandar = df_marketing[mnt_colums].mean(), df_marketing[mnt_colums].std()

print("Media de las columnas de gasto:", media)
print("Desviación estándar de las columnas de gasto:", desviacion_estandar)

```

### <font size=3 color="#5E5E5E"> <i class="fa fa-pencil-square-o" aria-hidden="true" style="color:#5E5E5E"></i> Ejercicio 41</font> 

Repetir el ejercicio anterior, pero utilizando `DataFrame.apply()`. Para ello, la función pasada como argumento debe devolver una tupla por cada columna.



```python
media, desviacion_estandar = df_marketing[mnt_colums].apply(np.mean), df_marketing[mnt_colums].apply(np.std)

print("Media de las columnas de gasto:", media)
print("Desviación estándar de las columnas de gasto:", desviacion_estandar)
```

### <font size=3 color="#5E5E5E"> <i class="fa fa-pencil-square-o" aria-hidden="true" style="color:#5E5E5E"></i> Ejercicio 42</font> 

Utilizar ahora `DataFrame.apply()` para obtener la media de las columnas en `mnt_colums` para cada fila.


```python
media =  df_marketing.apply(lambda x: x.mean() if x.name in mnt_colums else None)

print("Media de las columnas de gasto:", media)
```

### <font size=3 color="#5E5E5E"> <i class="fa fa-pencil-square-o" aria-hidden="true" style="color:#5E5E5E"></i> Ejercicio 44</font> 

Calcular el *Z-score* de la columna `Income` de cada cliente. Es decir, cuántas desviaciones estándar por encima o debajo de la media están los ingresos.

```python
df_marketing["Z_Score"] = (df_marketing["MntTotal"] - df_marketing["MntTotal"].mean()) / df_marketing["MntTotal"].std()
df_marketing["Z_Score"] 
```

### <font size=3 color="#5E5E5E"> <i class="fa fa-pencil-square-o" aria-hidden="true" style="color:#5E5E5E"></i> Ejercicio 45</font> 

Calcular el porcentaje de dinero que gasta cada cliente (suma de las columnas en `mnt_columns`) con respecto a sus ingresos. Utilizar para ello `DataFrame.apply()`

```python
df_marketing["Porcentaje_Gasto"] = df_marketing.apply(lambda row: (row["MntTotal"] / row["Income"]) * 100 if row["Income"] > 0 else 0, axis=1)
df_marketing["Porcentaje_Gasto"] 
```

### <font size=3 color="#5E5E5E"> <i class="fa fa-pencil-square-o" aria-hidden="true" style="color:#5E5E5E"></i> Ejercicio 46</font> 

Calcular la media y desviación de los del 50% de los clientes que tienen un mayor salario.

```python
salarios = df_marketing["Income"].sort_values(ascending=False)

top_50 = salarios.head(len(salarios) // 2)

media = top_50.mean()
desviacion = top_50.std()

print("Media:", media)
print("Desviación estándar:", desviacion)
```
