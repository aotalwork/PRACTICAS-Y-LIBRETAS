```python
import pandas as pd
import numpy as np
pd.__version__
```

En la mayoría de los casos, los `DataFrame` se crean a partir de la lectura de datos, y las `Series` se obtienen mediante el acceso a columnas (o filas). Sin embargo, en algunos casos puede ser necesario construir objetos a partir de datos que ya están en memoria: listas, tuplas y diccionarios. Veremos cómo se asignan las etiquetas y cómo se determinan los tipos de datos.

---
##  <font color="#5E5E5E"> 1. Construcción programática *Series*  </font>


Es posible crear objetos de tipo `Series` directamente, y de distintos modos, a partir de estructuras estándar de _Python_ y _Numpy_.


### <font color="#5E5E5E"> Construcción a partir de una secuencia de elementos</font>

Al construir una `Series` a partir de una lista, cada elemento de la lista aporta un valor. Si no indicamos `index`, pandas crea un `RangeIndex` con etiquetas `0, 1, 2, ...`. En este caso las etiquetas coinciden con las posiciones, pero son conceptos distintos.


```python
equipos = ['Real Madrid', 'Manchester United', 'Milán']
pd.Series(equipos)         # Construye una serie a partir de una lista. Por defecto, se indexa con enteros
```

Igual que al leer una columna textual, pandas 3.0 infiere aquí el dtype `str`. El tipo se determina a partir de los valores de la lista.

### <font color="#5E5E5E"> Construcción a partir de una colección y un índice </font>

El argumento `index` permite establecer las etiquetas del índice. Al construir la serie a partir de una lista, se debe proporcionar una etiqueta por valor, y estas se asocian por posición. Las etiquetas deben ser objetos hashables, como cadenas o números, y pueden repetirse.

```python
pd.Series(['Real Madrid', 'Manchester United', 'Milán'],  # Datos
           index=['España', 'Inglaterra', 'Italia'])       # Índice
```

### <font color="#5E5E5E"> Construcción a partir de un diccionario</font>

También se puede construir la serie a partir de un diccionario. En ese caso, las claves corresponden a los índices, y los valores a los elementos de la `Series`. 

```python
equipos = {'España': 'Real Madrid',
           'Inglaterra': 'Manchester United',
           'Italia': 'Milán',
           'Francia': 'PSG',
           'Alemania': 'Bayern Munich' }          # Diccionario

pd.Series(equipos)                                # Crea la serie a partir del diccionario
```

Para este caso, `index` define las etiquetas que conformarán el nuevo índice y su orden. Para cada etiqueta, *pandas* busca la clave correspondiente en el diccionario. Las entradas correspondientes a etiquetas no solicitadas se omiten y las etiquetas que no existen reciben un valor ausente. 

```python
s_equipos = pd.Series(equipos, index=['España', 'Inglaterra', 'Italia','Portugal'])
s_equipos
```

Es posible dar un nombre a la serie, y al índice de la serie. 

```python
s_equipos.name='Equipo'
s_equipos.index.name = 'País'
s_equipos
```

##  <font color="#5E5E5E"> 2. Construcción programática *DataFrame*  </font>


Existen _múltiples formas_ de crear un objeto `DataFrame` a partir de otros objetos como colecciones, diccionarios, `Series`, etc. Éstas son útiles, por ejemplo,  a la hora de construir el `DataFrame`
 a partir de datos procedentes de fuentes heterogéneas. 
<br>

### <font color="#5E5E5E"> Construcción a partir de una lista o colección de tuplas</font>



Se hace mediante el método `DataFrame.from_records()`. Cada elemento de la lista representa una fila. Si no se proporcionan las columnas e índice, en ambos casos se utilizan enteros.

```python
ventas = [('Álvaro', 22.5, 'Queso'),
         ('Benito', 14.5, 'Vino'),
         ('Fernando', 50, 'Jamón'),
         ('Martín', 20.0, 'Aceite'),
         ('Hernán',np.nan, 'Azafrán')]

columnas = ['Nombre', 'Precio', 'Producto']
indice =['Tienda 1', 'Tienda 1', 'Tienda 2', 'Tienda 3','Tienda 3']

df_ventas = pd.DataFrame.from_records(ventas, columns=columnas, index=indice)
df_ventas
```

```python
df_ventas.dtypes
```

El constructor general `pd.DataFrame()` también acepta los argumentos `index` y `columns`.

```python
pd.DataFrame(ventas, index=indice, columns=columnas)
```


### <font color="#5E5E5E"> Construcción a partir de diccionarios </font>

El método `DataFrame.from_dict()`  construye un `DataFrame` a partir de un diccionario. Mediante el parámetro `orient`, permite especificar si cada entrada del diccionario corresponde a una fila, o a una columna.

```python
ventas = {'Nombre': ['Álvaro', 'Benito', 'Fernando','Martín','Hernán'],
          'Precio': [22.5, 14.5, 50.0, 20.0, pd.NA],
          'Producto':['Queso', 'Vino', 'Jamón', 'Aceite','Azafrán']}

pd.DataFrame.from_dict(ventas, orient="index")
#pd.DataFrame.from_dict(ventas, orient="columns")
```


De igual manera, el método `DataFrame.to_dict()` permite exportar un `DataFrame` a un diccionario. El parámetro `orient` acepta siete valores: `dict`, `list`, `series`, `split`, `tight`, `records`e `index` (estrictamente hablando, algunos de ellos no devuelven un diccionario).


```python
df_ventas.reset_index(names="Tienda").to_dict(orient="index")
#df_ventas.reset_index(names="Tienda").to_dict(orient="records") # Devuelve una lista de diccionarios
```

El constructor acepta directamente un diccionario con el formato anterior (es equivalente a `pd.DataFrame.from_dict(orient="columns"))`.

```python
pd.DataFrame(ventas, index=indice)
```

Aunque lo ___recomendable es utilizar las dos funciones descritas anteriormente___, la construcción de `DataFrame` es muy versátil, y acepta multitud de opciones que, en alguna ocasión, pueden ser útiles. En este ejemplo, cada uno de los diccionarios representa una fila del `DataFrame`. Las claves representan el nombre de la columna, y los valores el valor correspondiente de la columna para la fila. 

```python
compra_1 =  {'Nombre': 'Álvaro', 'Producto': 'Queso', 'Precio': 22.50}
compra_2 =  {'Nombre': 'Benito', 'Producto': 'Vino', 'Precio': 14.50}
compra_3 =  {'Nombre': 'Fernando','Producto': 'Jamón', 'Precio': 50.00}
compra_4 =  {'Nombre': 'Martín', 'Producto': 'Aceite', 'Precio': 20.00}
compra_5 =  {'Nombre': 'Hernán', 'Producto': 'Azafrán'} # Deja a NaN los campos para los que no se proporciona valor.

df_ventas = pd.DataFrame([compra_1, compra_2, compra_3, compra_4, compra_5], 
                          index=['Tienda 1', 'Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 3'])
df_ventas
```

Cuando las columnas se proporcionan como `Series`, pandas alinea los valores por las etiquetas de sus índices.  Sin un índice explícito, el resultado incluye la unión de las etiquetas; donde falta un dato aparece un valor ausente

```python
nombres = pd.Series({"Tienda 1": "Álvaro", "Tienda 2": "Benito"})
precios = pd.Series({"Tienda 2": 14.5, "Tienda 1": 22.5, "Tienda 3": 50.0})

df_alineado = pd.DataFrame({"Nombre": nombres, "Precio": precios})
df_alineado
```
